# Data Visualizations - Top Indian Places to Visit

Comprehensive visualizations of the Indian places dataset including distributions, comparisons, and geographic insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

df = pd.read_csv('../data/cleaned_top_indian_places_to_visit.csv')
print("Dataset loaded successfully!")
print(f"Dataset shape: {df.shape}")
```

## 1. Google Review Ratings Distribution

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.hist(df['google_review_rating'], bins=20, color='purple', edgecolor='black', alpha=0.7)
plt.xlabel('Google Review Rating', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Google Review Ratings', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Entrance Fee Distribution

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.hist(df['entrance_fee_in_inr'], bins=30, color='coral', edgecolor='black', alpha=0.7)
plt.xlabel('Entrance Fee (INR)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Entrance Fees', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Time Needed to Visit Distribution

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.hist(df['time_needed_to_visit_in_hrs'], bins=25, color='lightgreen', edgecolor='black', alpha=0.7)
plt.xlabel('Time Needed (hours)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Time Needed to Visit Places', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Top 15 Places by Number of Google Reviews

```python
top_reviews = df.nlargest(15, 'number_of_google_review_in_lakhs')[['name', 'number_of_google_review_in_lakhs']].sort_values('number_of_google_review_in_lakhs')
fig, ax = plt.subplots(figsize=(12, 8))
plt.barh(top_reviews['name'], top_reviews['number_of_google_review_in_lakhs'], color='skyblue', edgecolor='black')
plt.xlabel('Number of Reviews (in lakhs)', fontsize=12)
plt.title('Top 15 Places by Number of Google Reviews', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Places Distribution by Zone

```python
zone_counts = df['zone'].value_counts().sort_values()
fig, ax = plt.subplots(figsize=(10, 6))
plt.barh(zone_counts.index, zone_counts.values, color='mediumpurple', edgecolor='black')
plt.xlabel('Number of Places', fontsize=12)
plt.title('Distribution of Places by Zone', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
for i, v in enumerate(zone_counts.values):
    plt.text(v + 0.3, i, str(v), va='center')
plt.tight_layout()
plt.show()
```

## 6. Top 10 Places by Type

```python
type_counts = df['type'].value_counts().head(10)
fig, ax = plt.subplots(figsize=(12, 6))
bars = plt.bar(range(len(type_counts)), type_counts.values, color='teal', edgecolor='black', alpha=0.7)
plt.xticks(range(len(type_counts)), type_counts.index, rotation=45, ha='right')
plt.ylabel('Number of Places', fontsize=12)
plt.title('Top 10 Place Types by Count', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(type_counts.values):
    plt.text(i, v + 0.5, str(v), ha='center', va='bottom')
plt.tight_layout()
plt.show()
```

## 7. Best Time to Visit Distribution

```python
best_time = df['best_time_to_visit'].value_counts()
fig, ax = plt.subplots(figsize=(10, 6))
colors_time = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
bars = plt.bar(best_time.index, best_time.values, color=colors_time, edgecolor='black', alpha=0.8)
plt.ylabel('Number of Places', fontsize=12)
plt.title('Best Time to Visit Distribution', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height)}', ha='center', va='bottom')
plt.tight_layout()
plt.show()
```

## 8. Scatter Plot: Google Rating vs Number of Reviews

```python
fig, ax = plt.subplots(figsize=(10, 6))
scatter = plt.scatter(df['google_review_rating'], df['number_of_google_review_in_lakhs'], 
                     c=df['entrance_fee_in_inr'], cmap='viridis', s=100, alpha=0.6, edgecolors='black')
plt.xlabel('Google Review Rating', fontsize=12)
plt.ylabel('Number of Reviews (in lakhs)', fontsize=12)
plt.title('Google Rating vs Number of Reviews', fontsize=14, fontweight='bold')
cbar = plt.colorbar(scatter)
cbar.set_label('Entrance Fee (INR)', fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 9. Box Plot: Google Rating by Place Type (Top 8 Types)

```python
top_8_types = df['type'].value_counts().head(8).index
df_top_types = df[df['type'].isin(top_8_types)]

fig, ax = plt.subplots(figsize=(12, 6))
sns.boxplot(data=df_top_types, x='type', y='google_review_rating', palette='Set2')
plt.xlabel('Place Type', fontsize=12)
plt.ylabel('Google Review Rating', fontsize=12)
plt.title('Distribution of Ratings by Top 8 Place Types', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 10. Top 10 States by Number of Places

```python
top_states = df['state'].value_counts().head(10).sort_values()
fig, ax = plt.subplots(figsize=(12, 7))
colors_states = plt.cm.Spectral(np.linspace(0, 1, len(top_states)))
plt.barh(top_states.index, top_states.values, color=colors_states, edgecolor='black')
plt.xlabel('Number of Places', fontsize=12)
plt.title('Top 10 States by Number of Places', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
for i, v in enumerate(top_states.values):
    plt.text(v + 0.3, i, str(v), va='center')
plt.tight_layout()
plt.show()
```

## 11. Average Entrance Fee by Significance Category

```python
avg_fee_by_sig = df.groupby('significance')['entrance_fee_in_inr'].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(11, 6))
bars = plt.bar(range(len(avg_fee_by_sig)), avg_fee_by_sig.values, 
               color='gold', edgecolor='black', alpha=0.8)
plt.xticks(range(len(avg_fee_by_sig)), avg_fee_by_sig.index, rotation=45, ha='right')
plt.ylabel('Average Entrance Fee (INR)', fontsize=12)
plt.title('Average Entrance Fee by Significance Category', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(avg_fee_by_sig.values):
    plt.text(i, v + 2, f'₹{v:.0f}', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 12. Average Rating by Zone

```python
avg_rating_by_zone = df.groupby('zone')['google_review_rating'].mean().sort_values()
fig, ax = plt.subplots(figsize=(10, 6))
bars = plt.barh(avg_rating_by_zone.index, avg_rating_by_zone.values, color='lightcoral', edgecolor='black')
plt.xlabel('Average Google Review Rating', fontsize=12)
plt.title('Average Google Review Rating by Zone', fontsize=14, fontweight='bold')
plt.xlim(4.0, 4.6)
for i, v in enumerate(avg_rating_by_zone.values):
    plt.text(v + 0.01, i, f'{v:.2f}', va='center')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 13. Entrance Fee vs Time Needed (Scatter Plot)

```python
fig, ax = plt.subplots(figsize=(11, 7))
scatter = plt.scatter(df['entrance_fee_in_inr'], df['time_needed_to_visit_in_hrs'], 
                     c=df['google_review_rating'], cmap='plasma', s=150, alpha=0.6, edgecolors='black', linewidth=0.5)
plt.xlabel('Entrance Fee (INR)', fontsize=12)
plt.ylabel('Time Needed to Visit (hours)', fontsize=12)
plt.title('Relationship: Entrance Fee vs Time Needed to Visit', fontsize=14, fontweight='bold')
cbar = plt.colorbar(scatter)
cbar.set_label('Google Review Rating', fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 14. Free vs Paid Places Distribution

```python
free_paid = df['entrance_fee_in_inr'].apply(lambda x: 'Free' if x == 0 else 'Paid').value_counts()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Pie chart
colors_fp = ['#90EE90', '#FFB6C1']
ax1.pie(free_paid.values, labels=free_paid.index, autopct='%1.1f%%', colors=colors_fp, 
        startangle=90, textprops={'fontsize': 12, 'weight': 'bold'})
ax1.set_title('Free vs Paid Places Distribution', fontsize=12, fontweight='bold')

# Bar chart
ax2.bar(free_paid.index, free_paid.values, color=colors_fp, edgecolor='black', alpha=0.8)
ax2.set_ylabel('Number of Places', fontsize=11)
ax2.set_title('Count: Free vs Paid Places', fontsize=12, fontweight='bold')
ax2.grid(axis='y', alpha=0.3)
for i, v in enumerate(free_paid.values):
    ax2.text(i, v + 2, str(v), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

print(f"Free Places: {free_paid['Free']}")
print(f"Paid Places: {free_paid['Paid']}")
```

## 15. Average Time Needed by Zone

```python
avg_time_by_zone = df.groupby('zone')['time_needed_to_visit_in_hrs'].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(10, 6))
bars = plt.bar(range(len(avg_time_by_zone)), avg_time_by_zone.values, 
               color='orange', edgecolor='black', alpha=0.8)
plt.xticks(range(len(avg_time_by_zone)), avg_time_by_zone.index, rotation=45, ha='right')
plt.ylabel('Average Time (hours)', fontsize=12)
plt.title('Average Time Needed to Visit by Zone', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(avg_time_by_zone.values):
    plt.text(i, v + 0.05, f'{v:.2f}h', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 16. Airport Availability by Zone

```python
airport_by_zone = df[df['airport_with_50km_radius'] == 'Yes'].groupby('zone').size()
no_airport = df[df['airport_with_50km_radius'] == 'No'].groupby('zone').size()

fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(len(airport_by_zone))
width = 0.35

bars1 = plt.bar(x - width/2, airport_by_zone.values, width, label='With Airport', color='skyblue', edgecolor='black')
bars2 = plt.bar(x + width/2, no_airport.values, width, label='No Airport', color='salmon', edgecolor='black')

plt.xlabel('Zone', fontsize=12)
plt.ylabel('Number of Places', fontsize=12)
plt.title('Airport Availability (50km radius) by Zone', fontsize=14, fontweight='bold')
plt.xticks(x, airport_by_zone.index, rotation=45, ha='right')
plt.legend(fontsize=11)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 17. Weekly Off Day Distribution

```python
weekly_off = df['weekly_off'].value_counts()
fig, ax = plt.subplots(figsize=(11, 6))
colors_wo = plt.cm.Set3(np.linspace(0, 1, len(weekly_off)))
bars = plt.bar(range(len(weekly_off)), weekly_off.values, color=colors_wo, edgecolor='black', alpha=0.8)
plt.xticks(range(len(weekly_off)), weekly_off.index, rotation=45, ha='right')
plt.ylabel('Number of Places', fontsize=12)
plt.title('Weekly Off Day Distribution', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(weekly_off.values):
    plt.text(i, v + 2, str(v), ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 18. Average Rating by Best Time to Visit

```python
avg_rating_by_time = df.groupby('best_time_to_visit')['google_review_rating'].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(11, 6))
bars = plt.bar(range(len(avg_rating_by_time)), avg_rating_by_time.values, 
               color='lightsteelblue', edgecolor='black', alpha=0.8)
plt.xticks(range(len(avg_rating_by_time)), avg_rating_by_time.index, fontsize=11)
plt.ylabel('Average Google Review Rating', fontsize=12)
plt.title('Average Google Review Rating by Best Time to Visit', fontsize=14, fontweight='bold')
plt.ylim(4.0, 4.6)
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(avg_rating_by_time.values):
    plt.text(i, v + 0.01, f'{v:.2f}', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 19. Establishment Year Distribution

```python
df_est = df[df['establishment_year'] != 'Unknown'].copy()
df_est['establishment_year'] = pd.to_numeric(df_est['establishment_year'], errors='coerce')
df_est = df_est.dropna(subset=['establishment_year'])

fig, ax = plt.subplots(figsize=(13, 6))
plt.hist(df_est['establishment_year'], bins=30, color='mediumpurple', edgecolor='black', alpha=0.7)
plt.xlabel('Establishment Year', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Place Establishment Years', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Oldest Place: {int(df_est['establishment_year'].min())}")
print(f"Newest Place: {int(df_est['establishment_year'].max())}")
print(f"Median Year: {int(df_est['establishment_year'].median())}")
```
